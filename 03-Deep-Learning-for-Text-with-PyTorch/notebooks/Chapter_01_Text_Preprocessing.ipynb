## 1.1 Text Processing Pipeline

Raw text must go through a **pipeline of transformations** before it can be used in a model:

```
Raw Text → Tokenization → Stop Word Removal → Stemming → Encoding → Model
```

## 1.2 Tokenization

Tokenization splits text into individual **tokens** (words, subwords, or characters).

Using `torchtext`:

````python
# Install if needed: pip install torchtext
from torchtext.data.utils import get_tokenizer

tokenizer = get_tokenizer("basic_english")

text = "I am reading a book now. I love to read books!"
tokens = tokenizer(text)

print("Original:", text)
print("Tokens:", tokens)
print("Count:", len(tokens))
````

## 1.3 Stop Word Removal

Stop words are very common words ("a", "the", "and") that add little meaning. Removing them:
- Reduces vocabulary size
- Focuses the model on meaningful content

````python
import nltk
nltk.download('stopwords', quiet=True)
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

tokens = ["i", "am", "reading", "a", "book", "now", ".",
          "i", "love", "to", "read", "books", "!"]

filtered_tokens = [token for token in tokens
                   if token.lower() not in stop_words and token.isalpha()]

print("Before:", tokens)
print("After:",  filtered_tokens)
````

## 1.4 Stemming

Stemming reduces words to their **root form**:
- "reading" → "read"
- "books" → "book"
- "running" → "run"

````python
from nltk.stem import PorterStemmer
nltk.download('punkt', quiet=True)

stemmer = PorterStemmer()

filtered_tokens = ['reading', 'love', 'read', 'books']
stemmed = [stemmer.stem(token) for token in filtered_tokens]

print("Before stemming:", filtered_tokens)
print("After stemming: ", stemmed)
````

## 1.5 Vocabulary Building

````python
from collections import Counter

# Build vocabulary from a corpus
corpus = [
    "I love deep learning",
    "PyTorch is a great deep learning framework",
    "Natural language processing with deep learning"
]

# Tokenize all documents
all_tokens = []
for doc in corpus:
    all_tokens.extend(tokenizer(doc))

# Count frequencies
freq = Counter(all_tokens)
print("Word frequencies:", freq.most_common(10))

# Build word-to-index mapping (reserve 0 for <PAD>, 1 for <UNK>)
vocab = {'<PAD>': 0, '<UNK>': 1}
for word, count in freq.most_common():
    if count >= 1:  # Minimum frequency threshold
        vocab[word] = len(vocab)

print(f"\nVocabulary size: {len(vocab)}")
print("Sample entries:", dict(list(vocab.items())[:10]))
````

## Summary

| Step | Tool | Purpose |
|---|---|---|
| Tokenization | `torchtext` | Split text into tokens |
| Stop word removal | `nltk` | Remove low-information words |
| Stemming | `nltk.PorterStemmer` | Reduce words to root form |
| Vocabulary | `Counter` | Map words to indices |
