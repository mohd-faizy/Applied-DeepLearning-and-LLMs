## 1.1 What Are Large Language Models (LLMs)?

**LLM**: An AI system able to understand and generate human-like text for addressing various complex language tasks.

**Popular LLMs:** GPT, BERT, LLaMA, Gemini, Claude, and more.

### Key Characteristics

| Feature | Detail |
|---|---|
| Architecture | Deep learning, most commonly **Transformers** |
| Pattern capture | Learn complex patterns in text data |
| NLP tasks | Text generation, summarization, translation, QA |
| "Large" | Very deep networks with billions of trainable weights |
| Training data | Vast amounts of text from the internet, books, etc. |

## 1.2 LLM Development Lifecycle

```
1. Data Ingestion & Preparation
      ↓
2. Model Architecture Design
      ↓
3. Pre-training on large corpora
      ↓
4. Fine-tuning on task-specific data
      ↓
5. Evaluation, Deployment & Monitoring
```

## 1.3 Hugging Face: The LLM Hub

[Hugging Face](https://huggingface.co/) is a community hub with:
- Thousands of **pre-trained LLMs** ready to use
- Large **datasets** for training
- Tools for fine-tuning, evaluation, and deployment

Core library: `transformers`

````python
# Install Hugging Face transformers
# pip install transformers

from transformers import pipeline

# Quick sentiment analysis with a pre-trained LLM
sentiment_pipeline = pipeline("sentiment-analysis")

texts = [
    "I love this product, it's amazing!",
    "This is the worst experience I've ever had.",
    "It's okay, nothing special."
]

results = sentiment_pipeline(texts)
for text, result in zip(texts, results):
    print(f"Text: {text}")
    print(f"  → Label: {result['label']}, Score: {result['score']:.3f}\n")
````

````python
# Text generation with GPT-2
generator = pipeline('text-generation', model='gpt2')

prompt = "The future of artificial intelligence is"
output = generator(
    prompt,
    max_length=60,
    num_return_sequences=2,
    temperature=0.8
)

for i, out in enumerate(output):
    print(f"Generation {i+1}:\n{out['generated_text']}\n")
````

## 1.4 Tasks Supported by Pre-trained LLMs

| Task | Pipeline Name | Description |
|---|---|---|
| Sentiment Analysis | `sentiment-analysis` | Positive/Negative classification |
| Text Generation | `text-generation` | Continues a prompt |
| Summarization | `summarization` | Condenses long text |
| Translation | `translation` | Translates between languages |
| Question Answering | `question-answering` | Answers from context |
| Fill Mask | `fill-mask` | Predicts masked tokens |

````python
# Summarization example
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

long_text = """
Artificial intelligence (AI) is intelligence demonstrated by machines, as opposed to the
natural intelligence displayed by animals including humans. AI research has been defined
as the field of study of intelligent agents, which refers to any system that perceives its
environment and takes actions that maximize its chance of achieving its goals.
"""

summary = summarizer(long_text, max_length=50, min_length=20, do_sample=False)
print("Summary:", summary[0]['summary_text'])
````
