## 1.1 What Are Large Language Models?

An **LLM** is an AI system able to understand and generate human-like text for complex language tasks. Popular examples include GPT, BERT, LLaMA, T5, and BART.

### Key Characteristics
- Usually built on deep learning architectures, most commonly transformers.
- Capture complex patterns in text data.
- Support text generation, summarization, translation, question answering, classification, intent recognition, NER, and code generation.
- Are called **large** because they contain many trainable weights and are trained on vast text datasets.

> **Note:** Model size is only one part of quality. Data quality, architecture, training objective, alignment, and evaluation all matter.


## 1.2 LLM Development Lifecycle

| Stage | Purpose |
|---|---|
| Data ingestion and preparation | Collect, clean, filter, deduplicate, and format text data |
| Model architecture design | Choose encoder, decoder, or encoder-decoder architecture and model size |
| Pre-training | Learn general language patterns from large corpora |
| Fine-tuning | Adapt the model to a task, domain, or preference style |
| Evaluation | Measure quality, safety, robustness, and limitations |
| Deployment, monitoring, maintenance | Serve the model and track drift, latency, cost, and quality |

### Section Summary
- LLM development is an end-to-end lifecycle.
- Evaluation and monitoring continue after deployment.


## 1.3 Hugging Face: The LLM Hub

[Hugging Face](https://huggingface.co/) is a community hub of pretrained LLMs, datasets, tokenizers, and evaluation tools.
The `pipeline()` function provides a high-level API that loads a suitable model and tokenizer for a task.

### Hello World: Sentiment Classification


```python
# If needed, install dependencies first:
# pip install transformers torch sentencepiece sacremoses

from transformers import pipeline

sentiment_classifier = pipeline("text-classification")
outputs = sentiment_classifier("""Dear seller, I got very impressed with the fast
                                delivery and careful packaging of my order. Great
                                experience overall, thank you!""")
print(outputs)
```

## 1.4 Tasks LLMs Can Perform

| Language Task | Goal | Common Architecture |
|---|---|---|
| Text classification | Assign labels such as sentiment or topic | Encoder-only |
| Text generation | Continue or complete text | Decoder-only |
| Text summarization | Produce a shorter version preserving key meaning | Encoder-decoder |
| Question answering | Extract or generate an answer | Encoder-only, encoder-decoder, decoder-only |
| Translation | Convert source language text to target language text | Encoder-decoder |
| Code generation | Generate code from requirements | Decoder-only |
| Intent recognition | Detect the purpose behind an utterance | Encoder-only |
| Named entity recognition | Identify and classify entities | Encoder-only |


### Text Classification: Sentiment Analysis


```python
from transformers import pipeline

llm = pipeline("text-classification")
text = "Walking amid Gion's Machiya wooden houses was a mesmerizing experience"
outputs = llm(text)
print(outputs[0]["label"])
```

### Text Generation


```python
from transformers import pipeline

llm = pipeline("text-generation")
prompt = "The Gion neighborhood in Kyoto is famous for"
outputs = llm(prompt, max_length=100)
print(outputs[0]["generated_text"])
```

### Text Summarization


```python
from transformers import pipeline

llm = pipeline("summarization", model="facebook/bart-large-cnn")
long_text = """Walking amid Gion's Machiya wooden houses is a mesmerizing experience. The beautifully
preserved structures exuded an old-world charm that transports visitors back in time, making them feel
like they had stepped into a living museum. The glow of lanterns lining the narrow streets add to the
enchanting ambiance, making each stroll a memorable journey through Japan's rich cultural history."""
outputs = llm(long_text, max_length=60, clean_up_tokenization_spaces=True)
print(outputs[0]["summary_text"])
```

### Question Answering


```python
from transformers import pipeline

llm = pipeline("question-answering")
context = "Walking amid Gion's Machiya wooden houses was a mesmerizing experience."
question = "What are Machiya houses made of?"
outputs = llm(question=question, context=context)
print(outputs["answer"])
```

### Language Translation


```python
from transformers import pipeline

llm = pipeline("translation_en_to_es", model="Helsinki-NLP/opus-mt-en-es")
text = "Walking amid Gion's Machiya wooden houses was a mesmerizing experience."
outputs = llm(text, clean_up_tokenization_spaces=True)
print(outputs[0]["translation_text"])
```

## 1.5 The Transformer Architecture

A **transformer** is a deep learning architecture for text processing, understanding, and generation. It is the foundation behind models such as BERT, GPT, T5, and BART.

### Transformer Characteristics
- No recurrent architecture is required.
- Tokens are handled simultaneously.
- Attention mechanisms and positional encoding capture long-range dependencies and order.
- Parallelism enables efficient training and inference.

### Original Transformer
The original architecture contains two stacks: an **encoder** and a **decoder**. Each stack repeats layers containing attention mechanisms, feed-forward sublayers, normalization, residual connections, and dropout.


```python
import torch
import torch.nn as nn

d_model = 512
n_heads = 8
num_encoder_layers = 6
num_decoder_layers = 6

model = nn.Transformer(
    d_model=d_model,
    nhead=n_heads,
    num_encoder_layers=num_encoder_layers,
    num_decoder_layers=num_decoder_layers,
)
print(model)
```

## 1.6 Transformer Architecture Families

| Type | Language Tasks | Notable Models |
|---|---|---|
| Encoder-decoder | Translation, summarization | T5, BART |
| Encoder-only | Text classification, extractive QA | BERT |
| Decoder-only | Text generation, generative QA | GPT |

## Chapter Summary
