## 7. Troubleshooting Playbook
### Symptom to Action
- OOM -> lower sequence length, enable checkpointing, reduce batch size
- Diverging loss -> lower LR, inspect noisy labels
- Hallucinations -> improve training quality and add retrieval grounding
- Slow training -> optimize dataloader and attention kernels

```python
def recommend_fix(symptom: str):
    symptom = symptom.lower()
    if "oom" in symptom:
        return "Reduce max_seq_length and batch size; enable gradient checkpointing."
    if "loss" in symptom:
        return "Lower learning rate and inspect label quality."
    if "halluc" in symptom:
        return "Improve factual data quality and add retrieval augmentation."
    return "Start with logs: GPU memory, LR schedule, gradient norms, and sample outputs."

for s in ["OOM at step 80", "Loss increasing", "Hallucination in finance answer"]:
    print(s, "->", recommend_fix(s))
```
