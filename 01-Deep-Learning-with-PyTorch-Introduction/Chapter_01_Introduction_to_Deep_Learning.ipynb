## 1.1 What is Deep Learning?

Deep learning is everywhere — it powers **language translation**, **self-driving cars**, **medical diagnostics**, and **chatbots**. It works on multiple data types: images, text, and audio.

| Traditional ML | Deep Learning |
|---|---|
| Relies on hand-crafted feature engineering | Enables automatic feature learning from raw data |

Deep learning is a **subset of machine learning**, inspired by the connections in the human brain. Models require a **large amount of data** to train effectively.

## 1.2 PyTorch: A Deep Learning Framework

PyTorch is:
- One of the most popular deep learning frameworks
- The framework used in many published deep learning research papers
- Intuitive and user-friendly — shares many concepts with NumPy

PyTorch also has domain-specific extensions:
- **`torchvision`** → image data
- **`torchaudio`** → audio data
- **`torchtext`** → text data

```python
# Importing PyTorch and related packages
import torch
import torchvision
import torchaudio
import torchtext

print(f"PyTorch version: {torch.__version__}")
```

## 1.3 Tensors: The Building Blocks of Neural Networks

Tensors are **multidimensional representations** of data — similar to NumPy arrays but with GPU acceleration support.

There are two main ways to create tensors:
1. **From a Python list** using `torch.tensor()`
2. **From a NumPy array** using `torch.from_numpy()`

```python
import torch
import numpy as np

# Load from list
lst = [[1, 2, 3], [4, 5, 6]]
tensor = torch.tensor(lst)
print("Tensor from list:\n", tensor)

# Load from NumPy array
np_array = np.array([[7, 8, 9], [10, 11, 12]])
np_tensor = torch.from_numpy(np_array)
print("\nTensor from NumPy:\n", np_tensor)
```

## 1.4 Tensor Attributes

Every tensor has key attributes:
- **`.shape`** — dimensions of the tensor
- **`.dtype`** — the data type (e.g., `torch.float32`)
- **`.device`** — where the tensor lives (`cpu` or `cuda`)

```python
lst = [[1, 2, 3], [4, 5, 6]]
tensor = torch.tensor(lst)

# Shape
print("Shape:", tensor.shape)

# Data type
print("Data type:", tensor.dtype)

# Device
print("Device:", tensor.device)

# Tensor type check
print("Is CUDA available:", torch.cuda.is_available())
```
