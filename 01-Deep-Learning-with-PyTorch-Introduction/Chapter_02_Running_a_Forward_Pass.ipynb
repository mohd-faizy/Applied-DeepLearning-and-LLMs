## 2.1 What Is a Forward Pass?

In a neural network, a **forward pass** means:
1. Input data is passed (propagated) forward through the network
2. Computations are performed at each layer
3. Outputs of each layer are passed to the next layer
4. The output of the final layer is called the **prediction**

Forward passes are used for both **training** and **inference (prediction)**.

### Possible Output Types

| Task | Output |
|---|---|
| Binary classification | Single probability between 0 and 1 |
| Multiclass classification | Distribution of probabilities summing to 1 |
| Regression | Continuous numerical value |

## 2.2 Training Loop: Forward + Backward Pass

During training, we run a full loop:
1. **Forward pass** — propagate input through the network
2. **Compare outputs** to true labels (compute loss)
3. **Backpropagation** — compute gradients and update model weights/biases
4. **Repeat** until the model produces useful outputs

## 2.3 Binary Classification Forward Pass

```python
import torch
import torch.nn as nn

# Create input data of shape 5x6 (5 samples, 6 features each)
input_data = torch.tensor([
    [-0.4421,  1.5207,  2.0607, -0.3647,  0.4691,  0.0946],
    [-0.9155, -0.0475, -1.3645,  0.6336, -1.9520, -0.3398],
    [ 0.7406,  0.2432,  1.6763, -0.8511,  0.1123, -0.0633],
    [-1.6630, -0.0718, -0.1285,  0.5396, -0.0288, -0.8622],
    [-0.7413,  1.7920, -0.0883, -0.6685,  0.4745, -0.4245]
])

# Create a binary classification model using nn.Sequential
model = nn.Sequential(
    nn.Linear(6, 4),   # First linear layer: 6 inputs -> 4 outputs
    nn.Linear(4, 1),   # Second linear layer: 4 inputs -> 1 output
    nn.Sigmoid()       # Sigmoid activation: squashes output to [0, 1]
)

# Pass input data through model (forward pass)
output = model(input_data)
print("Output shape:", output.shape)
print("Predictions (probabilities):\n", output)
```

## 2.4 Multiclass Classification Forward Pass

```python
# Multiclass classification: 5 samples, 6 features, 3 classes
model_multiclass = nn.Sequential(
    nn.Linear(6, 4),
    nn.Linear(4, 3),    # 3 output neurons for 3 classes
    nn.Softmax(dim=-1)  # Softmax: outputs sum to 1
)

output_multi = model_multiclass(input_data)
print("Multiclass output shape:", output_multi.shape)
print("Class probability distributions:\n", output_multi)
```

## Summary

| Pass | Purpose |
|---|---|
| Forward pass | Compute predictions from input |
| Backward pass | Compute gradients, update weights |
| Training loop | Alternates forward + backward passes repeatedly |
