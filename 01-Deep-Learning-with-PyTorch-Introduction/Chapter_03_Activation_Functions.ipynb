## 3.1 Limitations of Sigmoid and Softmax

Both **Sigmoid** and **Softmax** activation functions share a critical flaw:

- Their **gradients approach zero** for very low or very high input values
- This causes the function to **saturate**
- During backpropagation, multiplying by near-zero gradients causes the **vanishing gradients problem**
- Vanishing gradients make it very hard for early layers to learn

```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
import numpy as np

# Visualize sigmoid saturation
x = torch.linspace(-10, 10, 200)
sigmoid = nn.Sigmoid()
y_sigmoid = sigmoid(x).detach().numpy()

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.plot(x.numpy(), y_sigmoid)
plt.title('Sigmoid Function')
plt.xlabel('x')
plt.ylabel('σ(x)')
plt.grid(True)

# Derivative (gradient)
grad = y_sigmoid * (1 - y_sigmoid)
plt.subplot(1, 2, 2)
plt.plot(x.numpy(), grad)
plt.title('Sigmoid Gradient (approaches 0 at extremes)')
plt.xlabel('x')
plt.ylabel("σ'(x)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

## 3.2 ReLU — Rectified Linear Unit

ReLU is the most commonly used activation function in hidden layers.

$$f(x) = \max(x, 0)$$

- For **positive inputs**: output equals the input
- For **negative inputs**: output is zero
- Overcomes the vanishing gradients problem (gradient is always 1 for positive x)

```python
# ReLU in PyTorch
relu = nn.ReLU()

x = torch.linspace(-5, 5, 200)
y_relu = relu(x).detach().numpy()

plt.figure(figsize=(6, 4))
plt.plot(x.numpy(), y_relu, color='red', label='ReLU')
plt.title('ReLU Activation Function')
plt.xlabel('x')
plt.ylabel('ReLU(x)')
plt.legend()
plt.grid(True)
plt.show()

# Example usage in a model
print("ReLU on [-3, -1, 0, 1, 3]:", relu(torch.tensor([-3.0, -1.0, 0.0, 1.0, 3.0])))
```

## 3.3 Leaky ReLU

Leaky ReLU addresses the **dying ReLU problem** (neurons that always output zero):

- For **positive inputs**: behaves like ReLU
- For **negative inputs**: multiplies by a small slope coefficient (default `0.01`)
- Ensures gradients for negative inputs are **never exactly zero**

```python
# Leaky ReLU in PyTorch
leaky_relu = nn.LeakyReLU(negative_slope=0.05)

x = torch.linspace(-5, 5, 200)
y_leaky = leaky_relu(x).detach().numpy()

plt.figure(figsize=(6, 4))
plt.plot(x.numpy(), y_relu, label='ReLU', linestyle='--')
plt.plot(x.numpy(), y_leaky, label='Leaky ReLU (slope=0.05)', color='green')
plt.title('ReLU vs Leaky ReLU')
plt.xlabel('x')
plt.legend()
plt.grid(True)
plt.show()
```

## 3.4 Neural Network Architecture Deep Dive

### Layers Are Made of Neurons

**Linear (fully-connected) layers**: every neuron is connected to every neuron in the previous layer.

Each neuron:
- Computes a weighted linear combination of all previous layer's outputs
- Contains **N + 1** learnable parameters (N weights + 1 bias)

### Layer Naming Convention

| Name | Meaning |
|---|---|
| Input layer | Receives raw features |
| Hidden layers | Intermediate transformations |
| Output layer | Final prediction |

### Tweaking Hidden Layers
- **Input layer** size is fixed = number of features
- **Output layer** size is fixed = number of classes / outputs
- **Hidden layers** can be freely added and sized

```python
# Architecture example: configurable hidden layers
def build_model(n_features, n_hidden_layers, hidden_size, n_classes):
    layers = [nn.Linear(n_features, hidden_size), nn.ReLU()]
    for _ in range(n_hidden_layers - 1):
        layers += [nn.Linear(hidden_size, hidden_size), nn.ReLU()]
    layers.append(nn.Linear(hidden_size, n_classes))
    return nn.Sequential(*layers)

# Example: 10 features, 3 hidden layers, 64 units, 5 classes
model = build_model(n_features=10, n_hidden_layers=3, hidden_size=64, n_classes=5)
print(model)
```

## Summary

| Activation Function | Best Use |
|---|---|
| Sigmoid / Softmax | Output layer (classification) |
| ReLU | Hidden layers (most common) |
| Leaky ReLU | Hidden layers when neurons may "die" |
