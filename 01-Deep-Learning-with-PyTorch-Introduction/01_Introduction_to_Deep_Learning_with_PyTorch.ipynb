## 1.1 What Is Deep Learning?

Deep learning is a subset of machine learning that learns representations from data using neural networks. It powers language translation, self-driving cars, medical diagnostics, and chatbots.

| Traditional Machine Learning | Deep Learning |
|---|---|
| Often depends on hand-crafted feature engineering. | Learns useful features directly from raw data. |
| Works well with smaller, well-structured feature sets. | Often benefits from large datasets and high compute. |

> 🧠 **Key idea:** deep learning models are inspired by connections in the human brain and usually require a large amount of data.

## 1.3 Tensors: Building Blocks of Networks in PyTorch

A tensor is a multidimensional container for numerical data. Like NumPy arrays, tensors can represent scalars, vectors, matrices, and higher-dimensional data. PyTorch tensors can also live on GPUs.

The PDF introduces two common creation paths:
1. Load from a Python list with `torch.tensor(...)`.
2. Load from a NumPy array with `torch.from_numpy(...)`.


```python
import torch
import numpy as np

# Load from list
lst = [[1, 2, 3], [4, 5, 6]]
tensor = torch.tensor(lst)
print("Tensor from list:")
print(tensor)

# Load from NumPy array
array = [[7, 8, 9], [10, 11, 12]]
np_array = np.array(array)
np_tensor = torch.from_numpy(np_array)
print("\nTensor from NumPy array:")
print(np_tensor)
```

### Tensor Attributes

| Attribute | Meaning | Example |
|---|---|---|
| `.shape` | Tensor dimensions | `torch.Size([2, 3])` |
| `.dtype` | Data type of tensor elements | `torch.int64`, `torch.float32` |
| `.device` | Hardware location | `cpu`, `cuda:0` |

> ⚠️ Deep learning often benefits from GPUs because they offer parallel computation, faster training, and better performance for large tensor workloads.


```python
lst = [[1, 2, 3], [4, 5, 6]]
tensor = torch.tensor(lst)

print("Shape:", tensor.shape)
print("Data type:", tensor.dtype)
print("Device:", tensor.device)
print("CUDA available:", torch.cuda.is_available())
```

## 1.4 Getting Started with Tensor Operations

Basic tensor arithmetic is element-wise when tensor shapes are compatible. If dimensions do not align, PyTorch raises a shape mismatch error.


```python
# Compatible shapes
a = torch.tensor([[1, 1],
                  [2, 2]])

b = torch.tensor([[2, 2],
                  [3, 3]])

print("Addition:")
print(a + b)

print("\nSubtraction:")
print(a - b)

print("\nElement-wise multiplication:")
print(a * b)
```

```python
# Incompatible shapes
c = torch.tensor([[2, 2, 4],
                  [3, 3, 5]])

try:
    print(a + c)
except RuntimeError as error:
    print("RuntimeError:", error)
```

PyTorch tensors support many familiar array operations, including transposition, matrix multiplication, concatenation, and most NumPy-style operations.


```python
print("Transpose of a:")
print(a.T)

print("\nMatrix multiplication a @ b:")
print(a @ b)

print("\nConcatenate a and b along rows:")
print(torch.cat([a, b], dim=0))
```

## 1.5 Creating Our First Neural Network

A neural network layer transforms input data into output data. In PyTorch, `nn.Linear` applies a learned linear transformation:

$$y = XW^T + b$$

| Parameter | Role |
|---|---|
| `weight` | Multiplies the input features. |
| `bias` | Shifts the layer output. |

Initially, weights and biases are random and are not useful until they are tuned.


```python
import torch.nn as nn

# Create input_tensor with three features
input_tensor = torch.tensor([[0.3471, 0.4547, -0.2356]])

# Define our first linear layer
linear_layer = nn.Linear(in_features=3, out_features=2)

# Pass input through linear layer
output = linear_layer(input_tensor)
print("Output:")
print(output)

# Each linear layer has a .weight and .bias property
print("\nLayer weights:")
print(linear_layer.weight)

print("\nLayer bias:")
print(linear_layer.bias)
```

### Two-Layer Network Summary

| Item | Value |
|---|---|
| Input dimensions | `1 x 3` |
| `in_features` | `3` |
| `out_features` | `2` |
| Output dimensions | `1 x 2` |

Networks with only linear layers are called **fully connected** networks.


## 1.6 Stacking Layers with `nn.Sequential()`

`nn.Sequential` composes layers in order. The output size of one layer must match the input size of the next.


```python
# Create network with three linear layers
model = nn.Sequential(
    nn.Linear(10, 18),
    nn.Linear(18, 20),
    nn.Linear(20, 5)
)

input_tensor = torch.tensor([[-0.0014, 0.4038, 1.0305, 0.7521, 0.7489,
                             -0.3968, 0.0113, -1.3844, 0.8705, -0.9743]])

# Pass input_tensor to model to obtain output
output_tensor = model(input_tensor)
print("Input shape:", input_tensor.shape)
print("Output shape:", output_tensor.shape)
print(output_tensor)
```

> ⚠️ The output is a tensor of shape `1 x 5`, but it is not meaningful yet because the model parameters are still random.


## 1.7 Why Do We Need Activation Functions?

Stacking only linear layers still produces a linear relationship between inputs and outputs. Activation functions add **non-linearity**, allowing the model to learn more complex relationships.

| Activation | Typical Use |
|---|---|
| Sigmoid | Binary classification output probability |
| Softmax | Multiclass classification probability distribution |


### Meet the Sigmoid Function

For binary classification, sigmoid converts a pre-activation score into a value between 0 and 1.

| Sigmoid Output | Predicted Class |
|---|---|
| `> 0.5` | `1` |
| `<= 0.5` | `0` |


```python
import torch
import torch.nn as nn

input_tensor = torch.tensor([[6.0]])
sigmoid = nn.Sigmoid()
output = sigmoid(input_tensor)
print(output)
```

### Activation Function as the Last Layer

A sigmoid can be placed at the end of a neural network to produce a binary classification probability.

> 📝 Sigmoid as the final step after linear layers is equivalent in spirit to traditional logistic regression.


```python
model = nn.Sequential(
    nn.Linear(6, 4),  # First linear layer
    nn.Linear(4, 1),  # Second linear layer
    nn.Sigmoid()      # Sigmoid activation function
)

sample = torch.randn(1, 6)
print(model(sample))
```

### Getting Acquainted with Softmax

Softmax is used for multiclass classification. It returns a probability distribution:
- each element is bounded between 0 and 1;
- the output vector sums to 1;
- the highest value is the predicted class.

Example classes: `bird (0)`, `mammal (1)`, `reptile (2)`.


```python
import torch
import torch.nn as nn

# Create an input tensor
input_tensor = torch.tensor([[4.3, 6.1, 2.3]])

# Apply softmax along the last dimension
probabilities = nn.Softmax(dim=-1)
output_tensor = probabilities(input_tensor)

print(output_tensor)
print("Row sum:", output_tensor.sum(dim=-1))
```

`dim=-1` tells PyTorch to apply softmax over the input tensor's last dimension. This is the usual choice for batches shaped like `[batch_size, n_classes]`.
